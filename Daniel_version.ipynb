# Question 1: First Repeating Element
You are given an array of integers.

Write a function that returns the first repeating element in the array.

## Instructions:
* Traverse the array from left to right.
* Use a HashMap to track elements already seen.
* Return the first element that repeats.
* If no element repeats, return -1.

````python
def firstRepeatingElement(data) -> int:
    viewed_no = {}
    n = len(data)
    for i in range(n):
        if data[i] in viewed_no:
            return data[i]
        viewed_no[data[i]] = i 
    return -1

confirm = {
    1: [1, 1, 2, 3, 4,],
    -1: [1, 2, 3, 4],
    4: [1, 2, 3, 4, 4, 5, 5]
}

check = [firstRepeatingElement(col) == ans for ans, col in confirm.items()]

print(check)

````

# Question 2: Letter Combinations of a Phone Number

You are given a string containing digits from 0–9.

Return all possible letter combinations that the number could represent, in any order.

Each digit maps to letters as on a traditional phone keypad.

> **Note** that 1 does not map to any letters.

Digit Mapping:

```
2 → "abc"

3 → "def"

4 → "ghi"

5 → "jkl"

6 → "mno"

7 → "pqrs"

8 → "tuv"

9 → "wxyz"

0 → " " (space)

1 → "" (no letters)
```

````python
number_maps = {
    "0" : [" "],
    "1" : [""],
    "2" : ["a", "b", "c"],
    "3" : ["d", "e", "f"],
    "4" : ["g", "h", "i"],
    "5" : ['j', 'k', 'l'],
    '6' : ['m', 'n', 'o'],
    '7' : ['p', 'q', 'r', 's'],
    '8' : ['t', 'u', 'v'],
    '9' : ['w', 'x', 'y', 'z'],
    
}

# presses = {
#     "0" :1,
# "1" :1,
# "2" :3,
# "3" :3,
# "4" :3,
# "5" :3,
# '6' :3,
# '7' :4,
# '8' :3,
# '9' :4
# }

# typing library has a List element to run this 

# given a number 234
# adg aeh afi bdg beh bfi

# from itertools import pairwise

# def pair_combine(a, b):

from typing import List    

def letterCombinations(s: str):
    if not s:
        return []


    groups = []
    for d in s:
        letters = number_maps.get(d, "")
        if letters == "":
        
            return []
        groups.append(letters)
    
    res: List[str] = []

    def backtrack(index: int, path: List[str]):
        if index == len(groups):
            res.append("".join(path))
            return
        for ch in groups[index]:
            path.append(ch)
            backtrack(index + 1, path)
            path.pop()

    backtrack(0, [])
    return res

letterCombinations("123")
````

# Question 3: Stable Sort with Custom Comparator

You are given an array arr of length n.

Implement a stable sorting algorithm that sorts the array using a provided comparator function.

## Comparator
The comparator function cmp(a, b) returns:

- A negative integer if a < b
- 0 if a == b
- A positive integer if a > b

All comparisons must use this comparator.

## Constraints
- $1 \leq n \leq 10^6$
- Sorting must be stable
- The algorithm `must not use recursion`
- You may allocate only `one auxiliary array` of size n
- Time Complexity: O(n log n)
- Extra Space Complexity: O(n)

## Restrictions

You may not use:
- sorted()
- list.sort()
- Any library sorting function
- Sublist creation using slicing

````python
def comparator(a, b):
    if a < b:
        return -1
    elif a == b:
        return 0
    else:
        return 1

def swap(a, b):
    a, b = b, a 
    return a, b  
# [5, 2, 1]
def sortArray(c):
    n = len(c)
    for i in range(n):
        swapped = False
        for j in range(n - i - 1):

            if comparator(c[j], c[j + 1]) > 0:
                c[j], c[j + 1] = swap(c[j], c[j + 1])
                swapped = True
        if not swapped:
            return

sample = [5, 2, 1]
sortArray(sample)
print(sample)
````
